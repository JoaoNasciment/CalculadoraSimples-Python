```python

#CALCULADORA SIMPLES

print("Seja bem vindo ao meu projeto")
print("== CALCULADORA SIMPLES ==")
print("== VAMOS CALCULAR? ==")

num1 = int(input("Digite o primeiro numero:"))

#Qual tipo de conta fazer

print("Qual tipo de conta deseja fazer?")

modeloCalculo = input("(+ , - , * , / )")

#VERIFICA A OPERAÇÃO E REALIZA O CALCULO

num2 = int(input("Digite o segundo numero:"))

if modeloCalculo == "+":
  resultado = num1 + num2

elif modeloCalculo == "-":
  resultado = num1 - num2

elif modeloCalculo == "*":
  resultado = num1 * num2

#DIVISÃO POR ZERO
elif modeloCalculo == "/":
  if num2 == 0:
    resultado = "Não é possivel realizar uma divisão por zero!"

  else:
    resultado = num1 / num2

else:
  resultado = "Calculo invalido"

print(resultado)
```
